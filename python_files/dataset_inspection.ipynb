```python
from PIL import Image
import os

def check_images(folder_path, number_of_images=10):
    files = [
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print("Folder:", folder_path)
    print("Total images:", len(files))
    print()

    for filename in files[:number_of_images]:
        file_path = os.path.join(folder_path, filename)

        try:
            with Image.open(file_path) as img:
                print(
                    filename,
                    "-> Size:", img.size,
                    "| Mode:", img.mode,
                    "| Format:", img.format
                )

        except Exception as e:
            print(filename, "-> ERROR:", e)
```

```python
# Check Brain MRI - Normal

check_images(
    os.path.join(brain_path, "Normal")
)
```

```python
# Check Brain MRI - Tumor

check_images(
    os.path.join(brain_path, "Tumor")
)
```

```python
# Check all images for errors

def check_all_images(folder_path):
    files = [
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    total = len(files)
    valid = 0
    errors = []

    for filename in files:
        file_path = os.path.join(folder_path, filename)

        try:
            with Image.open(file_path) as img:
                img.verify()
            valid += 1

        except Exception as e:
            errors.append((filename, str(e)))

    print("Folder:", folder_path)
    print("Total images :", total)
    print("Valid images :", valid)
    print("Error images :", len(errors))

    if errors:
        print("\nProblem images:")
        for filename, error in errors:
            print(filename, "->", error)

    print("-" * 50)
```

```python
# Check all Brain MRI images

check_all_images(
    os.path.join(brain_path, "Normal")
)

check_all_images(
    os.path.join(brain_path, "Tumor")
)
```

```python
# Check all Lung CT images

check_all_images(
    os.path.join(ct_path, "Cancerous raw images-jpg")
)

check_all_images(
    os.path.join(ct_path, "Non-Cancerous raw images - jpg")
)
```

```python
import matplotlib.pyplot as plt
import random

def show_samples(folder_path, title, number_of_images=6):

    files = [
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    selected_files = random.sample(
        files,
        min(number_of_images, len(files))
    )

    plt.figure(figsize=(12, 6))

    for i, filename in enumerate(selected_files):

        file_path = os.path.join(folder_path, filename)

        image = Image.open(file_path)

        plt.subplot(2, 3, i + 1)
        plt.imshow(image)
        plt.title(filename)
        plt.axis("off")

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```

```python
# Show Brain MRI Normal samples

show_samples(
    os.path.join(brain_path, "Normal"),
    "Brain MRI - Normal"
)
```

```python
import os
import shutil
import random

random.seed(42)

# ==========================================
# FUNCTION TO SPLIT DATASET
# ==========================================

def split_dataset(source_folder, output_folder, class_names):
    
    # Create train, validation and test folders
    for split in ["train", "validation", "test"]:
        for class_name in class_names:
            folder = os.path.join(output_folder, split, class_name)
            os.makedirs(folder, exist_ok=True)

    for class_name in class_names:

        source_class_folder = os.path.join(source_folder, class_name)

        images = [
            f for f in os.listdir(source_class_folder)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ]

        # Shuffle images
        random.shuffle(images)

        total = len(images)

        # 70% train
        train_end = int(total * 0.70)

        # 15% validation
        validation_end = int(total * 0.85)

        train_images = images[:train_end]
        validation_images = images[train_end:validation_end]
        test_images = images[validation_end:]

        print("\nClass:", class_name)
        print("Total:", total)
        print("Train:", len(train_images))
        print("Validation:", len(validation_images))
        print("Test:", len(test_images))

        # Copy training images
        for image in train_images:
            source = os.path.join(source_class_folder, image)
            destination = os.path.join(
                output_folder, "train", class_name, image
            )
            shutil.copy2(source, destination)

        # Copy validation images
        for image in validation_images:
            source = os.path.join(source_class_folder, image)
            destination = os.path.join(
                output_folder, "validation", class_name, image
            )
            shutil.copy2(source, destination)

        # Copy test images
        for image in test_images:
            source = os.path.join(source_class_folder, image)
            destination = os.path.join(
                output_folder, "test", class_name, image
            )
            shutil.copy2(source, destination)

    print("\nDataset split completed!")
```

```python
brain_output = "processed_data/brain_mri"

brain_classes = [
    "Normal",
    "Tumor"
]

split_dataset(
    brain_path,
    brain_output,
    brain_classes
)
```

```python
# ==========================================
# SPLIT LUNG CT DATASET
# ==========================================

ct_output = "processed_data/lung_ct"

ct_classes = [
    "Cancerous raw images-jpg",
    "Non-Cancerous raw images - jpg"
]

split_dataset(
    ct_path,
    ct_output,
    ct_classes
)
```

```python
# ==========================================
# CHECK PROCESSED DATASET
# ==========================================

def count_processed_dataset(dataset_path, classes):

    for split in ["train", "validation", "test"]:

        print("\n==========", split.upper(), "==========")

        for class_name in classes:

            folder = os.path.join(
                dataset_path,
                split,
                class_name
            )

            files = [
                f for f in os.listdir(folder)
                if f.lower().endswith((".jpg", ".jpeg", ".png"))
            ]

            print(class_name, ":", len(files))
```

```python
count_processed_dataset(
    brain_output,
    brain_classes
)
```

```python
count_processed_dataset(
    ct_output,
    ct_classes
)
```

```python
# ==========================================
# VERIFY TRAIN / VALIDATION / TEST DATA
# ==========================================

def verify_processed_dataset(dataset_path, class_names):

    print("\n======================================")
    print("Dataset:", dataset_path)
    print("======================================")

    for split in ["train", "validation", "test"]:

        print("\n---", split.upper(), "---")

        total = 0

        for class_name in class_names:

            folder = os.path.join(
                dataset_path,
                split,
                class_name
            )

            if not os.path.exists(folder):
                print(class_name, ": FOLDER NOT FOUND")
                continue

            files = [
                f for f in os.listdir(folder)
                if f.lower().endswith((".jpg", ".jpeg", ".png"))
            ]

            print(class_name, ":", len(files))
            total += len(files)

        print("Total:", total)
```

```python
verify_processed_dataset(
    brain_output,
    brain_classes
)
```

```python
verify_processed_dataset(
    ct_output,
    ct_classes
)
```
