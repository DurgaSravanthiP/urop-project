```python
import os

# Dataset paths
brain_path = "datasets/brain_mri/Brain MRI Images"
ct_path = "datasets/lung_ct/CT-Scan Images"

print("Brain MRI path exists:", os.path.exists(brain_path))
print("Lung CT path exists:", os.path.exists(ct_path))
```

```python
# Count Brain MRI images

print("========== BRAIN MRI DATASET ==========")

for category in ["Normal", "Tumor"]:

    path = os.path.join(brain_path, category)

    files = [
        f for f in os.listdir(path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print(category, ":", len(files), "images")


# Count Lung CT images

print("\n========== LUNG CT DATASET ==========")

categories = [
    "Cancerous raw images-jpg",
    "Non-Cancerous raw images - jpg"
]

for category in categories:

    path = os.path.join(ct_path, category)

    files = [
        f for f in os.listdir(path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print(category, ":", len(files), "images")
```

```python
from PIL import Image
import os

def check_images(folder_path, number_of_images=10):
    files = [
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print("Folder:", folder_path)
    print("Total images:", len(files))
    print()

    for filename in files[:number_of_images]:
        file_path = os.path.join(folder_path, filename)

        try:
            with Image.open(file_path) as img:
                print(
                    filename,
                    "-> Size:", img.size,
                    "| Mode:", img.mode,
                    "| Format:", img.format
                )

        except Exception as e:
            print(filename, "-> ERROR:", e)
```

```python
# Check Brain MRI - Normal

check_images(
    os.path.join(brain_path, "Normal")
)
```
